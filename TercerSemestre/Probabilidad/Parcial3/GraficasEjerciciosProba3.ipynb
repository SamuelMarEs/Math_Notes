```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-np.pi, np.pi, 500)
condiciones = [x < 0, (x >= 0)&(x <= np.pi / 2), x > np.pi/2]
funciones = [lambda x: 0, lambda x: 1 - np.cos(x), lambda x: 1]

y = np.piecewise(x, condiciones, funciones)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x)", color="blue", linewidth=2)
plt.title("F(X)")
plt.show()
```

```python
x = np.linspace(-2, 2, 500)
condiciones = [x < 0, (x >= 0)&(x < 1 / 2), (x >= 1/2)&(x<1), x >= 1]
funciones = [lambda x: 0, lambda x: x*x, lambda x: 1-3*(1-x)/2, lambda x: 1]

y = np.piecewise(x, condiciones, funciones)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x)", color="blue", linewidth=2)
plt.title("F(X)")
plt.show()
```

```python
x = np.linspace(-6, 6, 700)
condiciones = [x < -1, (x >= -1)&(x <= 1), x > 1]
funciones = [lambda x: 1 / (4*x*x), lambda x: 1 / 4, lambda x: 1 / (4*x*x)]

y = np.piecewise(x, condiciones, funciones)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="f(x)", color="blue", linewidth=2)
plt.title("f(X)")
plt.show()
```
