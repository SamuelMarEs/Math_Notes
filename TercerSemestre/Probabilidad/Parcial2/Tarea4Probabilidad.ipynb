```python
import numpy as np
import matplotlib.pyplot as plt
```

Ejercicio  170

```python
x = [-2,-1,0,1,2]
fx = [1/8, 1/8, 1/2, 1/8, 1/8]

fig, ax = plt.subplots(figsize=(12,4))
ax.bar(x, fx, color='blue')
ax.set_title("Función de probabilidad, ej 170")
ax.set_xlabel("x")
ax.set_ylabel("f(x)")

plt.show()
```

Ejercicio 180

```python
from scipy.stats import geom
x = np.arange(0, 21, 1)
p = 1 / 6
fx = geom.pmf(x, p)

fig, ax = plt.subplots(figsize=(12,4))
ax.bar(x, fx)
ax.set_title('Ejercicio 180')
ax.set_xlabel("X")
ax.set_ylabel("f(x)")

plt.show
```

Ejercicio 191

```python
fig, ax = plt.subplots(figsize=(8, 5))

# Tramos horizontales: (x_ini, x_fin, altura)
tramos = [
    (-3, -1, 0),
    (-1,  1, 1/4),
    ( 1,  3, 1/2),
    ( 3,  5, 3/4),
    ( 5,  7, 1),
]

for x0, x1, y in tramos:
    ax.hlines(y, x0, x1, color='darkred', lw=2)

# Saltos verticales en las discontinuidades
for x, y0, y1 in [(-1, 0, 1/4), (1, 1/4, 1/2), (3, 1/2, 3/4), (5, 3/4, 1)]:
    ax.vlines(x, y0, y1, color='darkred', lw=2, linestyles='dashed')

# Puntos incluidos (rellenos) y excluidos (vacíos)
incluidos = [(-1, 1/4), (1, 1/2), (3, 3/4), (5, 1)]
excluidos = [(-1, 0),   (1, 1/4), (3, 1/2), (5, 3/4)]

for x, y in incluidos:
    ax.plot(x, y, 'o', color='darkred', markersize=8, zorder=5)
for x, y in excluidos:
    ax.plot(x, y, 'o', markerfacecolor='white',
            markeredgecolor='darkred', markersize=8, zorder=5)

ax.set_title('Ejercicio 191, CDF')
ax.set_xlabel('x'); ax.set_ylabel('F(x)')
ax.set_yticks([0, 1/4, 1/2, 3/4, 1])
ax.set_yticklabels(['0', '1/4', '1/2', '3/4', '1'])
ax.grid(alpha=0.3)
ax.set_ylim(-0.05, 1.1)
ax.set_xlim(-3, 7)
plt.show()
```

Ejercicio 192

```python
x = np.arange(1, 10)
fx = (1/2)**x

fig, ax = plt.subplots(figsize=(12,4))
ax.bar(x,fx)
ax.set_title("Ejercicio 192, PDF")
ax.set_xlabel("X")
ax.set_ylabel("f(x)")
```

```python
x = np.arange(1, 10)
Fx = 1 - (1/2)**x

fig, ax = plt.subplots(figsize=(12,4))
ax.step(x,Fx, color='darkred')
ax.set_title("Ejercicio 192, CDF")
ax.set_xlabel("X")
ax.set_ylabel("F(x)")
```

Ejercicio 194

```python
fig, ax = plt.subplots(figsize=(8, 5))

# Tramos horizontales: (x_ini, x_fin, altura)
tramos = [
    (-2, 0, 0),
    ( 0, 1, 1/5),
    ( 1, 2, 3/5),
    ( 2, 4, 1),
]

for x0, x1, y in tramos:
    ax.hlines(y, x0, x1, color='darkred', lw=2.5)

# Saltos verticales: (x, y_inferior, y_superior)
for x, y0, y1 in [(0, 0, 1/5), (1, 1/5, 3/5), (2, 3/5, 1)]:
    ax.vlines(x, y0, y1, color='darkred', lw=2, linestyles='dashed')

# Puntos incluidos (rellenos) y excluidos (vacíos)
incluidos = [(0, 1/5), (1, 3/5), (2, 1)]
excluidos = [(0, 0),   (1, 1/5), (2, 3/5)]

for x, y in incluidos:
    ax.plot(x, y, 'o', color='darkred', markersize=9, zorder=5)
for x, y in excluidos:
    ax.plot(x, y, 'o', markerfacecolor='white',
            markeredgecolor='darkred', markersize=9, zorder=5)

ax.set_title('Ejercicio 194')
ax.set_xlabel('x'); ax.set_ylabel('F(x)')
ax.set_yticks([0, 1/5, 3/5, 1])
ax.set_yticklabels(['0', '1/5', '3/5', '1'])
ax.set_xticks([-2, -1, 0, 1, 2, 3, 4])
ax.set_ylim(-0.05, 1.1)
ax.set_xlim(-2, 4)
ax.grid(alpha=0.3)
plt.show()
```
